### Exercise 1: Handle ZeroDivisionError
Write a Python program to handle a `ZeroDivisionError` exception when dividing a number by zero.


```python
try: 
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")    
```


### Exercise 2: Raise ValueError for Invalid Input
Write a Python program that prompts the user to input an integer and raises a `ValueError` exception if the input is not a valid integer.



```python
user_input = input("Enter an integer: ")

try:
    number = int(user_input)
    print(f"You entered integer: {number}")
except ValueError:
    print("Handled Error: Invalid input! The provided value is not an integer.")
```



### Exercise 3: Handle FileNotFoundError
Write a Python program that opens a file and handles a `FileNotFoundError` exception if the file does not exist.



```python
try:
    with open("non_existent_file.txt", "r") as file:
        content = file.read()
except FileNotFoundError:
    print("Error: The requested file could not be found.")
```



### Exercise 4: Raise TypeError for Non-Numerical Input
Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
val1 = input("Enter the first number: ")
val2 = input("Enter the second number: ")


def is_number(val):
    try:
        float(val)
        return True
    except ValueError:
        return False


if not (is_number(val1) and is_number(val2)):
    raise TypeError("Inputs must be numerical values.")
else:
    print(f"Sum: {float(val1) + float(val2)}")
```



### Exercise 5: Handle PermissionError
Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
try:
    
    with open("/etc/shadow", "w") as file:
        file.write("test")
except PermissionError:
    print(
        "Error: You do not have the required permissions to modify this file."
    )
```



### Exercise 6: Handle IndexError in List Operations
Write a Python program that executes an operation on a list and handles an `IndexError` exception if the index is out of range.




```python
my_list = [10, 20, 30]

try:
    element = my_list[5]
    print(element)
except IndexError:
    print("Error: List index out of range.")
```



### Exercise 7: Handle KeyboardInterrupt Exception
Write a Python program that prompts the user to input a number and handles a `KeyboardInterrupt` exception if the user cancels the input.



```python
try:
    num = input("Enter a number (or press Ctrl+C to cancel): ")
    print(f"You entered: {num}")
except KeyboardInterrupt:
    print("\nOperation cancelled by the user.")
```

Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
try:
    
    with open("sample_binary.bin", "r", encoding="utf-8") as file:
        content = file.read()
except UnicodeDecodeError:
    print("Error: Unable to decode file using UTF-8 encoding.")
except FileNotFoundError:
    print("File not found to demonstrate, but handler is configured.")
```
