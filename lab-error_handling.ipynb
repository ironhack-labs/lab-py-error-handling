### Exercise 1: Handle ZeroDivisionError
Write a Python program to handle a `ZeroDivisionError` exception when dividing a number by zero.


```python
def ZeroDivisionError(num1,num2):
    try:
        result = num1 / num2
        print("The result is:", result)
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
input1 = float(input("Enter the first number: "))
input2 = float(input("Enter the second number: "))
ZeroDivisionError(input1, input2)
```


### Exercise 2: Raise ValueError for Invalid Input
Write a Python program that prompts the user to input an integer and raises a `ValueError` exception if the input is not a valid integer.



```python
def ask_number(num):
    try:
        num = int(num)
        return print(f"Congratulations! You entered a valid integer: {num}")
    except ValueError:
        print("Error: Please enter a valid integer.")
ask_number(input("Enter an integer: "))
```



### Exercise 3: Handle FileNotFoundError
Write a Python program that opens a file and handles a `FileNotFoundError` exception if the file does not exist.



```python
def read_file(filename):
    try:
        with open (filename, 'r') as file:
            content = file.read()
            print(content)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
input1 = input("Enter the filename to read: ")
read_file(input1)
```



### Exercise 4: Raise TypeError for Non-Numerical Input
Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



### Exercise 5: Handle PermissionError
Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
def open_file(file_route):
    try:
        with open(file_route, 'r') as file:
            content = file.read()
            print(f"File content:\n{content}")
    except FileNotFoundError:
        print(f"Error: The file/Route: '{file_route}' was not found.")
    except PermissionError:
        print(f"Error: Permission denied to open the file '{file_route}'.")
open_file(input("Enter the file route to open: "))
```



### Exercise 6: Handle IndexError in List Operations
Write a Python program that executes an operation on a list and handles an `IndexError` exception if the index is out of range.




```python
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
def get_weekday(index):
    try:
        return print(f"The weekday at index {index} is: {weekdays[index-1]}")
    except IndexError:
        print(f"Error: Index {index} is out of range. Please enter a valid index between 1 and 7.")
index = int(input("Enter an index to get the corresponding weekday (1-7): "))
ask_number(index)
get_weekday(index)
```

Write a Python program that executes division and handles an `ArithmeticError` exception if there is an arithmetic error.



```python
def divide_numbers(num1, num2):
    try:
        result = float(num1) / float(num2)
        print(f"The result of dividing {num1} by {num2} is: {result}")
    except ArithmeticError as e:
        print(f"Error: {e}. Please enter valid numbers for division.")
    except ValueError:
        print("Value Error: Please enter valid numbers for division.")
divide_numbers(input("Enter the first number for division: "), input("Enter the second number for division: "))
```



### Exercise 9: Handle UnicodeDecodeError
Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
def open_file_endcode(file_route, encoding='utf-8'):
    try:
        with open(file_route, 'r', encoding=encoding) as file:
            content = file.read()
            print(f"File content:\n{content}")
    except UnicodeDecodeError:
        print(f"Error: Could not decode the file '{file_route}' with encoding '{encoding}'. Please check the file's encoding.")
    except FileNotFoundError:
        print(f"Error: The file/Route: '{file_route}' was not found.")
    except PermissionError:
        print(f"Error: Permission denied to open the file '{file_route}'.")
open_file_endcode(input("Enter the file route to open: "), input("Enter the encoding: "))
```



### Exercise 10: Handle AttributeError
Write a Python program that executes an operation on an object and handles an `AttributeError` exception if the attribute does not exist.



```python
def execute_operation(object):
    try:
        object.append(1)
    except AttributeError:
        print("Error: The object does not support the 'append' operation. Please provide a list or similar object.")
execute_operation(input("Enter an object: "))
```
