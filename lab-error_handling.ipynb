### Exercise 1: Handle ZeroDivisionError
Write a Python program to handle a `ZeroDivisionError` exception when dividing a number by zero.


```python
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print(result)
```


### Exercise 2: Raise ValueError for Invalid Input
Write a Python program that prompts the user to input an integer and raises a `ValueError` exception if the input is not a valid integer.



```python
def read_integer(value=None, prompt="Enter an integer: "):
    if value is None:
        value = input(prompt)
    try:
        return int(value)
    except ValueError as exc:
        raise ValueError("Input must be a valid integer.") from exc

try:
    print(read_integer(value="abc"))
except ValueError as e:
    print(e)
```



### Exercise 3: Handle FileNotFoundError
Write a Python program that opens a file and handles a `FileNotFoundError` exception if the file does not exist.



```python
def read_file_content(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError:
        print(f"File '{path}' does not exist.")
        return None

print(read_file_content("missing_file.txt"))
```



### Exercise 4: Raise TypeError for Non-Numerical Input
Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
def add_numbers(a, b):
    try:
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Both inputs must be numerical values.")
        return a + b
    except TypeError as exc:
        print(exc)
        return None

print(add_numbers("5", 2))
```



### Exercise 5: Handle PermissionError
Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
def open_file_safely(path):
    try:
        if path == "restricted.txt":
            raise PermissionError("Permission denied while opening the file.")
        with open(path, "r", encoding="utf-8") as file:
            return file.read()
    except PermissionError as exc:
        print(f"Permission error: {exc}")
        return None

print(open_file_safely("restricted.txt"))
```

Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
def read_text_with_decode(path):
    try:
        with open(path, "rb") as file:
            data = file.read()
        text = data.decode("utf-8")
        return text
    except UnicodeDecodeError:
        print("Encoding issue: the file is not valid UTF-8.")
        return None

# Example file content created in binary with invalid UTF-8
with open("bad_encoding.txt", "wb") as file:
    file.write(b"\xff\xfe\x00\x00")

print(read_text_with_decode("bad_encoding.txt"))
```

### Bonus Exercise 1: Handle Multiple Exceptions
Write a Python program that demonstrates handling multiple exceptions in one block.




```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero.")
    except TypeError:
        print("Both arguments must be numbers.")
    except ValueError:
        print("Value error encountered.")

print(divide(10, 2))
print(divide(10, 0))
print(divide("10", 2))
```



### Bonus Exercise 2: Create Custom Exception
Create a custom exception class and raise it in your code when certain conditions are met.




```python
class CustomValueError(Exception):
    """Raised when a custom validation rule fails."""


def validate_age(age):
    if age < 0:
        raise CustomValueError("Age cannot be negative.")
    return age

try:
    print(validate_age(-1))
except CustomValueError as exc:
    print(f"Custom exception: {exc}")
```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
import logging

logging.basicConfig(filename="error_log.txt", level=logging.ERROR, format="%(asctime)s:%(levelname)s:%(message)s")

try:
    value = 10 / 0
except ZeroDivisionError as exc:
    logging.error("ZeroDivisionError: %s", exc)
    print("Error logged to file.")
```



### Bonus Exercise 5: Retry Logic on Exception
Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
def risky_operation():
    import random
    if random.random() < 0.7:
        raise ValueError("Random failure")
    return "Success"

max_retries = 3
for attempt in range(1, max_retries + 1):
    try:
        result = risky_operation()
        print(result)
        break
    except ValueError as exc:
        print(f"Attempt {attempt} failed: {exc}")
        if attempt == max_retries:
            print("Maximum retries reached.")
```
