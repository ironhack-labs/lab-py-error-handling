### Exercise 1: Handle ZeroDivisionError
Write a Python program to handle a `ZeroDivisionError` exception when dividing a number by zero.


```python
# Your code here
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as e:
        print("ZeroDivisionError caught:", e)
        return None
    else:
        # runs only if NO exception was raised
        return result
    finally:
        # runs either way - typically used for cleanup
        pass


print("10 / 2 =", divide(10, 2))
print("10 / 0 =", divide(10, 0))

# Note that 10 / 0 raises ZeroDivisionError but 10.0 / 0 does too, while
# numpy would instead return inf with a warning. Plain Python always raises.
```


### Exercise 2: Raise ValueError for Invalid Input
Write a Python program that prompts the user to input an integer and raises a `ValueError` exception if the input is not a valid integer.



```python
# Your code here
# The lab says to prompt the user. input() blocks when the notebook is run
# top to bottom, so the prompts are simulated with a list of test values.
# The interactive version would be:  user_input = input("Enter an integer: ")

def parse_integer(user_input):
    try:
        value = int(user_input)
    except ValueError:
        # re-raise with a clearer message of our own
        raise ValueError("'%s' is not a valid integer" % user_input)
    return value


for test in ["42", "-7", "3.9", "abc", ""]:
    try:
        print("input %-6r -> %d" % (test, parse_integer(test)))
    except ValueError as e:
        print("input %-6r -> ValueError: %s" % (test, e))

# "3.9" fails too: int() will not silently truncate a string containing a
# decimal point, though int(3.9) on an actual float gives 3.
```

Write a Python program that opens a file and handles a `FileNotFoundError` exception if the file does not exist.



```python
# Your code here
def read_file(path):
    try:
        with open(path, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError as e:
        print("FileNotFoundError caught:", e)
        return None


# A file that does exist
print("README.md ->", (read_file("README.md") or "")[:40].replace("\n", " "), "...")
print()

# ...and one that does not
content = read_file("no_such_file.txt")
print("returned:", content)

# Using `with` matters here: it closes the file even if an exception is
# raised while reading, which a bare open() would not.
```

Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
# Your code here
# Prompts simulated again - see Exercise 2 for why.

def add_numbers(a, b):
    for value in (a, b):
        if not isinstance(value, (int, float)) or isinstance(value, bool):
            raise TypeError("expected a number, got %s: %r"
                            % (type(value).__name__, value))
    return a + b


for pair in [(3, 4), (2.5, 1.5), ("5", 2), (3, None), (True, 1)]:
    try:
        print("%-14r -> %s" % (pair, add_numbers(*pair)))
    except TypeError as e:
        print("%-14r -> TypeError: %s" % (pair, e))

# bool is excluded deliberately. In Python True is an int (isinstance(True,
# int) is True, and True + 1 == 2), so without that guard a boolean would
# slip through a numeric check unnoticed.
```

Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
# Your code here
import os

def open_file(path):
    try:
        with open(path, "r") as f:
            return f.read()
    except PermissionError as e:
        print("PermissionError caught:", e)
        return None
    except FileNotFoundError as e:
        print("FileNotFoundError caught:", e)
        return None


# Opening a DIRECTORY as if it were a file raises PermissionError on Windows
# (on Linux it raises IsADirectoryError, which is a different subclass of
# OSError - worth knowing if this notebook is run elsewhere).
print("trying to open the current directory as a file:")
open_file(os.getcwd())

print()
print("PermissionError is a subclass of OSError:",
      issubclass(PermissionError, OSError))
```

Write a Python program that executes an operation on a list and handles an `IndexError` exception if the index is out of range.




```python
# Your code here
def get_item(lst, index):
    try:
        return lst[index]
    except IndexError as e:
        print("IndexError caught: %s (list has %d items, asked for index %d)"
              % (e, len(lst), index))
        return None


colours = ["red", "green", "blue"]

print("index  2 ->", get_item(colours, 2))
print("index -1 ->", get_item(colours, -1))      # negative indexing is valid
print("index  5 ->", get_item(colours, 5))
print()

# A slice does NOT raise - it silently returns what it can:
print("colours[1:99] ->", colours[1:99])
print("which is why slicing hides out-of-range bugs that indexing exposes.")
```

Write a Python program that prompts the user to input a number and handles a `KeyboardInterrupt` exception if the user cancels the input.



```python
# Your code here
# KeyboardInterrupt is raised when the user presses Ctrl+C. It cannot be
# triggered by waiting here, so it is raised directly to show the handler.

def ask_for_number(simulate_interrupt=False):
    try:
        if simulate_interrupt:
            raise KeyboardInterrupt
        # the interactive version:  return int(input("Enter a number: "))
        return 42
    except KeyboardInterrupt:
        print("KeyboardInterrupt caught - input cancelled by the user")
        return None


print("normal   ->", ask_for_number(False))
print("cancelled->", ask_for_number(True))

print()
print("KeyboardInterrupt inherits from BaseException, not Exception:")
print("  issubclass(KeyboardInterrupt, Exception)    :",
      issubclass(KeyboardInterrupt, Exception))
print("  issubclass(KeyboardInterrupt, BaseException):",
      issubclass(KeyboardInterrupt, BaseException))
print()
print("That is deliberate. 'except Exception' does NOT catch it, so a broad")
print("handler cannot accidentally swallow Ctrl+C and leave a program that")
print("refuses to quit.")
```

Write a Python program that executes division and handles an `ArithmeticError` exception if there is an arithmetic error.



```python
# Your code here
def safe_divide(a, b):
    try:
        return a / b
    except ArithmeticError as e:
        print("ArithmeticError caught: %s: %s" % (type(e).__name__, e))
        return None


print("8 / 2   ->", safe_divide(8, 2))
print("8 / 0   ->", safe_divide(8, 0))
print()

# ArithmeticError is the PARENT class. Catching it also catches:
for cls in (ZeroDivisionError, OverflowError, FloatingPointError):
    print("  %-20s subclass of ArithmeticError: %s"
          % (cls.__name__, issubclass(cls, ArithmeticError)))

print()
# OverflowError, a different ArithmeticError, from the same handler:
import math
try:
    math.exp(1000)
except ArithmeticError as e:
    print("math.exp(1000) -> %s: %s" % (type(e).__name__, e))

# Catching the parent is convenient but less precise - the handler can no
# longer tell which failure occurred without inspecting type(e).
```

Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
# Your code here
# Create a file containing bytes that are not valid UTF-8, then read it as
# UTF-8 so the error actually happens rather than being described.
with open("bad_encoding.bin", "wb") as f:
    f.write(b"valid ascii, then an invalid byte: \xff\xfe and more text")

def read_as_utf8(path):
    try:
        with open(path, "r", encoding="utf-8") as f:
            return f.read()
    except UnicodeDecodeError as e:
        print("UnicodeDecodeError caught:")
        print("   codec   :", e.encoding)
        print("   position:", e.start)
        print("   reason  :", e.reason)
        return None


print("reading as utf-8:")
read_as_utf8("bad_encoding.bin")

print()
print("the same file read as latin-1:")
with open("bad_encoding.bin", "r", encoding="latin-1") as f:
    print("  ", repr(f.read()[:60]))

print()
print("latin-1 never raises - it maps every one of the 256 byte values to a")
print("character. That makes it a useful fallback, but it will happily")
print("produce nonsense rather than telling you the encoding was wrong.")

import os
os.remove("bad_encoding.bin")
```

Write a Python program that executes an operation on an object and handles an `AttributeError` exception if the attribute does not exist.



```python
# Your code here
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return "%s says woof" % self.name


def call_method(obj, method_name):
    try:
        method = getattr(obj, method_name)
    except AttributeError as e:
        print("AttributeError caught:", e)
        return None
    return method()


d = Dog("Rex")
print("bark()  ->", call_method(d, "bark"))
print("meow()  ->", call_method(d, "meow"))

print()
# getattr takes a default, which avoids the exception entirely:
print("getattr with a default:", getattr(d, "meow", "no such method"))
print("hasattr check          :", hasattr(d, "meow"))

print()
print("Three ways to handle a missing attribute: try/except, getattr with a")
print("default, or hasattr first. The exception is the right choice when the")
print("attribute is normally expected to exist and its absence is a genuine")
print("problem - not as routine flow control.")
```

### Bonus Exercise 1: Handle Multiple Exceptions
Write a Python program that demonstrates handling multiple exceptions in one block.




```python
# Your code here
def risky_operation(data, index, divisor):
    try:
        value = data[index]
        number = int(value)
        return number / divisor

    except (IndexError, KeyError) as e:
        # one handler for two related lookup failures
        print("  lookup failed -> %s: %s" % (type(e).__name__, e))
    except ValueError as e:
        print("  conversion failed -> ValueError: %s" % e)
    except ZeroDivisionError:
        print("  division failed -> ZeroDivisionError")
    except Exception as e:
        # catch-all LAST, so the specific handlers above get first refusal
        print("  unexpected -> %s: %s" % (type(e).__name__, e))
    return None


data = ["10", "abc", "20"]

for args in [(0, 2), (1, 2), (0, 0), (9, 2), (2, None)]:
    print("index=%s divisor=%s" % args)
    print("  result:", risky_operation(data, *args))

# Order matters. Python takes the FIRST matching handler, so specific
# exceptions must come before general ones - put `except Exception` at the
# top and nothing below it would ever run.
```

Create a custom exception class and raise it in your code when certain conditions are met.




```python
# Your code here
class InsufficientFundsError(Exception):
    """Raised when an account lacks the balance for a withdrawal."""

    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        self.shortfall = amount - balance
        super().__init__(
            "cannot withdraw %.2f from a balance of %.2f (short by %.2f)"
            % (amount, balance, self.shortfall))


class Account:
    def __init__(self, balance=0):
        self.balance = balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("withdrawal amount must be positive")
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance


acct = Account(100)

for amount in [30, 500, -5]:
    try:
        print("withdraw %6.2f -> balance %.2f" % (amount, acct.withdraw(amount)))
    except InsufficientFundsError as e:
        print("withdraw %6.2f -> %s" % (amount, e))
        print("                  shortfall available as e.shortfall: %.2f"
              % e.shortfall)
    except ValueError as e:
        print("withdraw %6.2f -> ValueError: %s" % (amount, e))

# Carrying balance, amount and shortfall as attributes is the point of a
# custom class. The caller can act on the numbers programmatically instead
# of parsing them back out of an error message.
```

Write a program that repeatedly prompts the user for valid input until they provide it, using exception handling to manage invalid inputs.



```python
# Your code here
# Prompts simulated - the real version would be:
#     raw = input("Enter your age: ")

def validate_age(inputs):
    """Keep asking until a valid age arrives, or the inputs run out."""
    for raw in inputs:
        try:
            age = int(raw)
            if age < 0:
                raise ValueError("age cannot be negative")
            if age > 130:
                raise ValueError("age %d is not plausible" % age)
        except ValueError as e:
            print("  rejected %-8r -> %s" % (raw, e))
            continue        # ask again
        else:
            print("  accepted %-8r -> %d" % (raw, age))
            return age
    print("  ran out of attempts")
    return None


print("attempt sequence: 'abc', '-5', '200', '34'")
result = validate_age(["abc", "-5", "200", "34"])
print()
print("final value:", result)

# The loop only exits via `return` in the else branch, so invalid input can
# never fall through. A `while True` with input() behaves identically - this
# version just takes its values from a list so the cell actually runs.
```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
# Your code here
import logging, os

LOG_FILE = "error_log.txt"
if os.path.exists(LOG_FILE):
    os.remove(LOG_FILE)

logging.basicConfig(
    filename=LOG_FILE,
    level=logging.ERROR,
    format="%(asctime)s | %(levelname)s | %(message)s",
    force=True,          # replace any handler a previous run installed
)

def divide_and_log(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        # exc_info=True writes the full traceback to the file
        logging.error("division failed: %s / %s", a, b, exc_info=True)
        return None
    except TypeError:
        logging.error("bad types: %r / %r", a, b, exc_info=True)
        return None


print("results:")
for a, b in [(10, 2), (1, 0), ("x", 2), (9, 3)]:
    print("  %r / %r -> %s" % (a, b, divide_and_log(a, b)))

logging.shutdown()

print()
print("--- contents of %s ---" % LOG_FILE)
with open(LOG_FILE, encoding="utf-8") as f:
    for line in f.read().splitlines()[:14]:
        print(line)

print()
print("Logging to a file rather than printing means the errors survive after")
print("the program exits, and the console stays readable for normal output.")
```

Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
# Your code here
import time

class FlakyService:
    """Fails the first `fail_times` calls, then succeeds."""

    def __init__(self, fail_times):
        self.calls = 0
        self.fail_times = fail_times

    def fetch(self):
        self.calls += 1
        if self.calls <= self.fail_times:
            raise ConnectionError("attempt %d failed" % self.calls)
        return "data received on attempt %d" % self.calls


def with_retry(operation, max_attempts=4, delay=0.1):
    for attempt in range(1, max_attempts + 1):
        try:
            return operation()
        except ConnectionError as e:
            print("  attempt %d/%d failed: %s" % (attempt, max_attempts, e))
            if attempt == max_attempts:
                print("  giving up after %d attempts" % max_attempts)
                raise
            wait = delay * (2 ** (attempt - 1))   # exponential backoff
            print("  retrying in %.2fs" % wait)
            time.sleep(wait)


print("service that fails twice then succeeds:")
print("  ->", with_retry(FlakyService(fail_times=2).fetch))

print()
print("service that always fails:")
try:
    with_retry(FlakyService(fail_times=99).fetch)
except ConnectionError as e:
    print("  propagated to the caller: %s" % e)

# Two things make this retry loop safe rather than dangerous:
#
#   max_attempts - without a limit a permanently broken service produces an
#   infinite loop.
#
#   backoff - the wait doubles each time (0.1, 0.2, 0.4s). Retrying
#   immediately and repeatedly just adds load to something already failing.
#
# Also note the bare `raise` on the final attempt: it re-raises the original
# exception with its traceback intact, rather than hiding the failure.
```
