### Exercise 1: Handle ZeroDivisionError
Write a Python program to handle a `ZeroDivisionError` exception when dividing a number by zero.


```python
try:
    result = 10/ 0
    print(result)
except ZeroDivisionError:
    print("You can't divide by zero")    
```


### Exercise 2: Raise ValueError for Invalid Input

```python
try:
    number = int(input("Please enter an integer"))
    print("You entered:", number)
except ValueError:
    print("That's not a valid integer")
```



### Exercise 4: Raise TypeError for Non-Numerical Input
Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
pre_weight = input("Weight before dialysis (kg)")
post_weight = input("Weight after dialysis (kg)")

try:
    pre = float(pre_weight)
    post = float(post_weight)
    removed = pre - post
    print("Fluid removed:", round(removed, 1), "kg")
except ValueError:
    raise TypeError("Both weights must be numbers!")
```

Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
with open("old_system_export.txt", "w", encoding="latin-1") as file:
    file.write("Patient: Frau Müller, Gewicht: 68.4 kg")

try:
    with open("old_system_export.txt", encoding="utf-8") as file:
        print(file.read())
except UnicodeDecodeError:
    print("This file uses a different encoding and cannot be read correctly.")
```



### Exercise 10: Handle AttributeError
Write a Python program that executes an operation on an object and handles an `AttributeError` exception if the attribute does not exist.



```python
class Patient:
    def __init__(self, name, dry_weight):
        self.name = name
        self.dry_weight = dry_weight

patient = Patient("Mrs. Klein", 64.5)

try:
    print("Dry weight:", patient.dry_weight)
    print("Shunt location:", patient.shunt_location)
except AttributeError:
    print("This information is not on the patient's record.")
```



### Bonus Exercise 2: Create Custom Exception
Create a custom exception class and raise it in your code when certain conditions are met.




```python
class LowBloodPressureError(Exception):
    pass

systolic = 85

try:
    if systolic < 90:
        raise LowBloodPressureError("Systolic below 90: stop ultrafiltration and inform the doctor.")
    print("Blood pressure OK:", systolic)
except LowBloodPressureError as alarm:
    print("ALARM:", alarm)
```



### Bonus Exercise 3: Validate User Input with Exception Handling
Write a program that repeatedly prompts the user for valid input until they provide it, using exception handling to manage invalid inputs.



```python
while True:
    try:
        weight = float(input("Enter the patient's weight before dialysis (kg):"))
        break
    except ValueError:
        print("That's not a valid weight. Please try again.")

print("Weight recorded:", weight, "kg")        
```



### Bonus Exercise 4: Log Errors to File
Modify your error handling to log errors to a text file instead of printing them to the console.



```python
try:
    fluid = int(input("Fluid to remove (ml): "))
    hours = int(input("Session time (hours): "))
    print("UF rate:", fluid / hours, "ml per hour")
except (ValueError, ZeroDivisionError) as error:
    with open("error_log.txt", "a") as log:
        log.write("Error: " + str(error) + "\n")
```



### Bonus Exercise 5: Retry Logic on Exception
Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
max_attempts = 3
recorded = False

for attempt in range(1, max_attempts + 1):
    try:
        systolic = int(input("Enter systolic blood pressure: "))
        print("Blood pressure recorded:", systolic)
        recorded = True
        break
    except ValueError:
        print("Invalid entry. Attempt", attempt, "of", max_attempts)

if recorded == False:
    print("Too many invalid entries. Please call a colleague to check.")
```
