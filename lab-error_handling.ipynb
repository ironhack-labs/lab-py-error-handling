Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
# get numerical input
def get_numerical_input(prompt):
    user_input =input(prompt)
    try:
        # Try converting the input to float
        return  float(user_input)
    except ValueError:
        # If conversion fails, raise a TypeError as requested
        raise TypeError(f"Invalid input: '{user_input}' is not a valid number.")

try:
    print("--- Number Validator ---")
    
    # input the numbers 
    num1 = get_numerical_input("enter the first number")
    num2 = get_numerical_input("enter the second number")

    # If successful, perform a simple operation
    print(f"\nSuccess! You entered {num1} and {num2}.")
    print(f"Their sum is: {num1 + num2}")

except TypeError as error:
# Catch and display the custom TypeError
    print(f"\nCaught an Exception: {error}")
```

```python
file_path = "protected_file.txt"

try:
    # Attempting to open a file with a strict encoding rules
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()
        print("\nSuccess! File content:")
        print(content)

except UnicodeDecodeError:
    # This block triggers if the file cannot be decoded using utf-8
    print(f"\nDecoding Error: Failed to read the file due to an encoding mismatch.")
```

Write a Python program that executes an operation on an object and handles an `AttributeError` exception if the attribute does not exist.



```python
class Cat:
    def __init__(self, name):
        self.name = name
    
    def bark(self):
        return f"{self.name} says Meo!"

# Create a Cat Object
my_pet = Cat("Cutie")

try:
    # 1. This will work fine
    print(my_pet.bark())
    
    # 2. This will raise an AttributeError because Cats don't fly!
    print("Attempting to make the Cat fly...")
    my_pet.fly() 

except AttributeError:
    # Catch and handle the missing attribute cleanly
    print(f"\nError Caught: That operation is not supported by this object.")
   
```

### Bonus Exercise 1: Handle Multiple Exceptions
Write a Python program that demonstrates handling multiple exceptions in one block.




```python
# Handling ZeroDivision Error and TypeError in one program
def calculate_division(a, b):
    try:
        # Perform division
        result = a / b
        print(f"Success! {a} divided by {b} is: {result}")
        
    except ZeroDivisionError:
        # Triggers if b is 0
        print("Math Error: Cannot divide by zero.")
        
    except TypeError:
        # Triggers if a or b is not a number (e.g., a string or a list)
        print("Type Error: Both inputs must be numbers.")

# --- Testing the program with different scenarios ---

print("--- Scenario 1: Normal Division ---")
calculate_division(10, 2)

print("\n--- Scenario 2: ZeroDivisionError ---")
calculate_division(10, 0)

print("\n--- Scenario 3: TypeError ---")
calculate_division(10, "two")  # "two" is a string, not a number

```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
# Age Validator
print("--- Age Validator (File Logging) ---")

while True:
    try:
        # Prompt the user for input
        user_input = input("enter your age")
        # Try to convert into integer
        age = int(user_input)

        # if conversion sucessful  break out of the loop
        break
    except ValueError:
    # Open 'log.txt'file in append mode ('a') to add the error to the file
        with open("log.txt", "a") as log_file:
            log_file.write(f"Error: Invalid input '{user_input}'\n")

# This code runs only after a valid input is received
print(f"\nSuccess! Your registered age is: {age}")
```

Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
# Set the starting number of tries allowed
attempts = 3

# Keep running the loop as long as the user has tries left
while attempts > 0:
    try:
         # Ask the user to type something in
        user_input = input("Enter a number: ")

         # Try to convert float
        number = float(user_input)

        print(f"Success! You entered: {number}")
        break  # Stops the program immediately on success

    except ValueError:
         # This code runs ONLY if the user typed text instead of a number
        attempts = attempts - 1  # Lose 1 try
        if attempts > 0:
            print(f"Invalid. You have {attempts} tries left.\n")
        else:
             # This runs if attempts hit 
            print("Game over! Out of attempts.")
```
