
### Exercise 2: Raise ValueError for Invalid Input
Write a Python program that prompts the user to input an integer and raises a `ValueError` exception if the input is not a valid integer.



```python
def read_integer(text):
    """Convert text to int or raise a clear ValueError."""
    try:
        return int(text)
    except ValueError:
        raise ValueError(f"'{text}' is not a valid integer")

# Simulated user input instead of input() so the notebook runs without interaction
for user_input in ["42", "abc"]:
    try:
        print("Valid integer:", read_integer(user_input))
    except ValueError as e:
        print("ValueError:", e)
```



### Exercise 4: Raise TypeError for Non-Numerical Input
Write a Python program that prompts the user to input two numbers and raises a `TypeError` exception if the inputs are not numerical.



```python
def add_numbers(a, b):
    """Add two numbers, raise TypeError if an input is not numeric."""
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Both inputs must be numbers")
    return a + b

for a, b in [(3, 4.5), (3, "x")]:
    try:
        print("Sum:", add_numbers(a, b))
    except TypeError as e:
        print("TypeError:", e)
```



### Exercise 5: Handle PermissionError
Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
# Running as root would ignore file permissions, so we raise the error ourselves
# to demonstrate the handling. In a normal user session open() on a protected file raises it.
def open_protected(path):
    raise PermissionError(f"No permission to open {path}")

try:
    open_protected("/root/secret.txt")
except PermissionError as e:
    print("PermissionError:", e)
```

Write a Python program that opens a file and handles a `UnicodeDecodeError` exception if there is an encoding issue.



```python
# Write bytes that are not valid UTF-8 and try to read them as UTF-8
with open("bad_encoding.txt", "wb") as f:
    f.write(b"\xff\xfe\xfa")

try:
    with open("bad_encoding.txt", encoding="utf-8") as f:
        f.read()
except UnicodeDecodeError as e:
    print("UnicodeDecodeError:", e.reason)
```

Write a Python program that demonstrates handling multiple exceptions in one block.




```python
def risky(x):
    return 10 / int(x)

for x in ["2", "0", "abc"]:
    try:
        print(risky(x))
    except (ZeroDivisionError, ValueError) as e:
        print(f"Handled {type(e).__name__}: {e}")
```



### Bonus Exercise 2: Create Custom Exception
Create a custom exception class and raise it in your code when certain conditions are met.




```python
class NegativeAgeError(Exception):
    """Raised when an age is below zero."""

def set_age(age):
    if age < 0:
        raise NegativeAgeError("Age cannot be negative")
    return age

try:
    set_age(-5)
except NegativeAgeError as e:
    print("Custom exception:", e)
```

Write a program that repeatedly prompts the user for valid input until they provide it, using exception handling to manage invalid inputs.



```python
# Simulated inputs: the first two are invalid, the third one is valid
attempts = iter(["abc", "-", "7"])

while True:
    user_input = next(attempts)  # replace with input("Enter an integer: ") in real use
    try:
        number = int(user_input)
        print("Accepted:", number)
        break
    except ValueError:
        print(f"'{user_input}' is invalid, try again")
```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
import logging

logging.basicConfig(filename="errors.log", level=logging.ERROR, format="%(asctime)s %(levelname)s %(message)s")

try:
    1 / 0
except ZeroDivisionError as e:
    logging.error("Division failed: %s", e)

print(open("errors.log").read())
```

Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
import random

random.seed(1)

def flaky_operation():
    if random.random() < 0.7:
        raise ConnectionError("temporary failure")
    return "success"

max_retries = 5
for attempt in range(1, max_retries + 1):
    try:
        print(f"Attempt {attempt}:", flaky_operation())
        break
    except ConnectionError as e:
        print(f"Attempt {attempt} failed: {e}")
else:
    print("All retries failed")
```
