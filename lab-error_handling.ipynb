

### Bonus Exercise 2: Create Custom Exception
Create a custom exception class and raise it in your code when certain conditions are met.




```python
# Your code here
class MyCustomException(Exception):
    def __init__(self, exception):
        self.exception = exception
        super().__init__(f'My custom Exception raised! Original error: {exception}')

a = [1,2,3,4]

try:
    a[10]
except IndexError as e:
    try:
        raise MyCustomException(e) from e
    except MyCustomException as custom_error:
        print(custom_error)
```

Write a program that repeatedly prompts the user for valid input until they provide it, using exception handling to manage invalid inputs.



```python
# Your code here
while True:
    a = input('Please enter a number')
    try:
        float(a)
    except Exception as e:
        print(f'Invalid input: \'{a}\', Please try again.')
        continue
    print('Your number is : ', a)
    break
```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
# Your code here
file = open('./log.txt', 'a')

while True:
    a = input('Please enter a number')
    try:
        float(a)
    except Exception as e:
        file.write(f'Invalid input: \'{a}\', Please try again.\n')
        continue
    print('Your number is : ', a)
    break
file.close()
```



### Bonus Exercise 5: Retry Logic on Exception
Implement retry logic for operations that could fail, allowing users to try again after encountering an error.



```python
# Your code here
max_retries = 3
attempts = 0

while attempts < max_retries:

    try:
        a = input('Please enter a number: ')
        number = float(a)

        print('Your number is:', number)
        break

    except ValueError:
        attempts += 1
        print(f'Invalid input! Attempts remaining: {max_retries - attempts}')

else:
    print('Maximum attempts reached!')
```
