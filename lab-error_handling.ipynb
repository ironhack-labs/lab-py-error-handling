Write a Python program that opens a file and handles a `PermissionError` exception if there is a permission issue.




```python
filename = "README.md"

try:
    with open(filename, "r") as f:
        content = f.read()
        print(content)
except PermissionError:
    print(f"Error: You don't have permission to access '{filename}'.")
```



### Exercise 9: Handle UnicodeDecodeError



```python
filename = "README.md"

try:
    with open(filename, "r", encoding="euc-kr") as f:
        content = f.read()
        print(content)
except UnicodeDecodeError:
    print(f"Error: Could not decode '{filename}' — it may not be UTF-8 encoded.")
```

Write a Python program that demonstrates handling multiple exceptions in one block.

Create a custom exception class and raise it in your code when certain conditions are met.




```python
class NotGraduatedError(Exception):
    """Raised when student doesnt graduate."""

def check_has_diploma(has_skills):
    if not has_skills:
        raise NotGraduatedError (f"Student has skills")

try: 
    check_has_diploma(False)
except NotGraduatedError as e:
    print(f"Error: {e}")
    
```

Modify your error handling to log errors to a text file instead of printing them to the console.



```python
def division(a,b):
    try:
        print(a/b)
    except ZeroDivisionError as e:
        print("Error: Cannot divide by zero.")
        with open("myfile.txt", "a") as f:
            f.write(f" {e}.\n")
    except TypeError as e:
        print("Error: Values need to be numeric")
        with open("myfile.txt", "a") as f:
            f.write(f" {e}.\n")
        
division(20,2)
division(20,0)
division("hh",2)
```
